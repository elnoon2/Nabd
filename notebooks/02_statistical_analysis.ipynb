# 02 - Statistical analysis

```python
import pandas as pd
from scipy import stats
import statsmodels.formula.api as smf
wait = pd.read_csv('../data/cleaned/wait_logs.csv')
```

```python
# Q2: peak vs off-peak (Welch)
peak = wait[wait.is_peak==1].wait_minutes; off = wait[wait.is_peak==0].wait_minutes
stats.ttest_ind(peak, off, equal_var=False)
```

```python
# Q2: specialties (Kruskal-Wallis)
stats.kruskal(*[g.wait_minutes for _, g in wait.groupby('specialty')])
```

```python
# Q4: multiple regression
smf.ols('wait_minutes ~ hour + patients_present', wait).fit().summary()
```
